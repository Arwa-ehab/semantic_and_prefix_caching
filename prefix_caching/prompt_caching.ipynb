```python
#Prompt that exceeds the 1024 token limit for testing purposes
system_prompt = f"""
You are an advanced HR assistant designed to answer employee questions about company policies. 
Your role is to provide clear, concise, and accurate answers using ONLY the information provided 
in the policy documentation below. If the answer cannot be found in the documentation, respond with:
"I cannot find this information in the company policy."

When answering:
- Be concise and professional.
- Cite the relevant policy section when possible.
- Do not invent information.
- Do not answer questions unrelated to company policy.

--------------------------------------
COMPANY HUMAN RESOURCES POLICY MANUAL
--------------------------------------

SECTION 1: COMPANY VALUES

Our company is committed to maintaining a respectful, inclusive, and productive work environment.
Employees are expected to treat colleagues with professionalism and mutual respect. Harassment,
discrimination, or inappropriate conduct is strictly prohibited.

SECTION 2: WORK HOURS

Standard working hours are 40 hours per week. Employees typically work from 9:00 AM to 5:00 PM,
Monday through Friday. Flexible work arrangements may be approved by managers depending on team
requirements.

Employees are expected to:
- Arrive on time
- Notify their manager if they will be late
- Log their working hours in the internal system

SECTION 3: REMOTE WORK POLICY

Employees may work remotely up to two days per week depending on their role and manager approval.
Remote work arrangements must be documented in the HR system.

Remote employees must:
- Remain reachable during working hours
- Attend required meetings
- Maintain productivity expectations

SECTION 4: ANNUAL LEAVE

Full-time employees are entitled to 21 days of paid annual leave per year.
Unused leave may be carried over to the following year up to a maximum of 5 days.

Leave requests must:
- Be submitted through the HR system
- Be approved by the direct manager
- Be requested at least 5 business days in advance unless urgent

SECTION 5: MATERNITY LEAVE

Employees are entitled to 16 weeks of paid maternity leave. Maternity leave can begin
up to two weeks before the expected due date.

Additional unpaid leave may be requested subject to HR approval.

SECTION 6: PATERNITY LEAVE

Employees are entitled to 5 working days of paid paternity leave following the birth
of a child.

SECTION 7: DRESS CODE

The company follows a business casual dress code. Acceptable clothing includes:

For men:
- Collared shirts
- Polo shirts
- Dress pants or chinos
- Closed shoes

For women:
- Blouses
- Dresses
- Dress pants
- Closed shoes or professional footwear

Clothing that is not permitted includes:
- Athletic wear
- Flip-flops
- Clothing with offensive graphics
- Excessively casual clothing

SECTION 8: CODE OF CONDUCT

Employees must maintain professional behavior in all company interactions.
Examples of misconduct include:

- Harassment
- Fraud
- Data misuse
- Disrespectful communication

Violations may result in disciplinary action including termination.

SECTION 9: INFORMATION SECURITY

Employees must follow company security policies including:

- Using strong passwords
- Locking computers when away from desk
- Not sharing confidential information
- Reporting suspicious activity

SECTION 10: PERFORMANCE REVIEWS

Employees receive performance reviews twice per year. Reviews evaluate:

- Work quality
- Collaboration
- Communication
- Initiative

Managers provide feedback and discuss development plans.

--------------------------------------

Use this policy documentation to answer employee questions accurately.
Always prioritize the policy text over assumptions.

If a question asks about HR policies, benefits, leave, conduct, or workplace rules,
search the policy text carefully before answering.
SECTION 11: EMPLOYEE BENEFITS

The company provides a comprehensive benefits package to eligible employees.
Benefits are designed to support employee health, financial security, and
overall well-being.

Benefits include:

- Health insurance coverage for employees and eligible dependents
- Dental insurance
- Vision insurance
- Life insurance
- Disability insurance
- Retirement savings plans
- Wellness programs

Employees become eligible for benefits after completing the onboarding
process and meeting employment eligibility requirements.

SECTION 12: HEALTH INSURANCE

The company offers health insurance plans that cover medical consultations,
hospitalization, prescription medications, and preventative care.

Employees may select from multiple plan options depending on their needs.

Coverage includes:

- Doctor visits
- Emergency care
- Hospital stays
- Prescription medication
- Preventive screenings

Employees are responsible for reviewing their coverage options during the
annual enrollment period.

SECTION 13: SICK LEAVE POLICY

Employees are entitled to paid sick leave when they are unable to work due to
illness or medical appointments.

Guidelines include:

- Employees should notify their manager as soon as possible.
- Sick leave may require medical documentation for extended absences.
- Sick leave is intended strictly for health-related reasons.

Abuse of sick leave may result in disciplinary action.



--------------------------------------------------

IMPORTANT INSTRUCTIONS FOR THE ASSISTANT

Always answer questions based strictly on the HR policy manual above.

When possible:

- Reference the relevant section number.
- Provide a concise explanation.
- Avoid unnecessary speculation.

If the policy does not contain the answer, respond with:

"I cannot find this information in the company policy."

"""
```

```python
from langchain_core.messages import ToolMessage,SystemMessage,HumanMessage,AIMessage
from langchain_openai.chat_models import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
model=ChatOpenAI(model="gpt-4.1",temperature=0, verbose=True)

messages = [SystemMessage(content=system_prompt)]

questions = [
    "What is the dress code policy?",
    "How many maternity leave weeks are allowed?",
    "How many annual leave days do employees get?",
    "What is the remote work policy?",
    "What benefits do employees receive?",
    "What is the sick leave policy?",
    "How does the grievance process work?",
    "What are the working hours?",
    "What is the travel reimbursement policy?",
    "What happens during employee onboarding?"
]

total_prompt_tokens = 0
total_cached_tokens = 0

for i, q in enumerate(questions):

    print(f"\n===== QUESTION {i+1} =====")
    print("User:", q)
    history=messages+[HumanMessage(content=q)]
    
    start_time = time.time()
    response = model.invoke(history,prompt_cache_key= "hr_policy_demo")
    end_time = time.time()

    # response_time approximation (non-streaming)
    response_time = end_time - start_time

    # Extract token usage
    token_usage = response.response_metadata['token_usage']
    prompt_tokens = token_usage['prompt_tokens']
    cached_tokens = token_usage['prompt_tokens_details']['cached_tokens']

    total_prompt_tokens += prompt_tokens
    total_cached_tokens += cached_tokens

    print("AI:", response.content)
    print("Prompt Tokens:", prompt_tokens)
    print("Cached Tokens:", cached_tokens)
    print("response_time:", round(response_time, 3), "seconds")

 

# Overall cache hit ratio
cache_hit_ratio = total_cached_tokens / total_prompt_tokens

print("\n==============================")
print("TOTAL PROMPT TOKENS:", total_prompt_tokens)
print("TOTAL CACHED TOKENS:", total_cached_tokens)
print("CACHE HIT RATIO:", round(cache_hit_ratio * 100, 2), "%")
```

```python
#Testing Gemini Explicit Caching
import time
from google import genai
from google.genai import types
from langchain.messages import HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI
#
client = genai.Client()

# Upload file
file = client.files.upload(file="D:\\ai tech talk\\caching\\rag_data\\ChapterSix_en.pdf")
while file.state.name == "PROCESSING":
    time.sleep(2)
    file = client.files.get(name=file.name)

# Create cache
model = "gemini-2.5-flash"
cache = client.caches.create(
    model=model,
    config=types.CreateCachedContentConfig(
        display_name="Cached Content",
        system_instruction=(
            "You are an expert content analyzer, and your job is to answer "
            "the user's query based on the file you have access to."
        ),
        contents=[file],
        ttl="300s",
    ),
)

# Query with LangChain
llm = ChatGoogleGenerativeAI(
    model=model,
    cached_content=cache.name,
)
message = HumanMessage(content="Summarize the main points of the content.")
llm.invoke([message])
```

```python
#Query with a question that is not in the file to see the cache in action
message = HumanMessage(content="who is the president of the United States?  ")
response=llm.invoke([message])
print(response)
token_usage = response.usage_metadata
prompt_tokens = token_usage['input_tokens']
cached_tokens = token_usage['input_token_details']['cache_read']
#Token usage should show some cached tokens for the second question, even though the answer is not in the file, because the model can still use the cached content to understand the context of the question and generate a response based on that understanding.
print("Prompt Tokens:", prompt_tokens)
print("Cached Tokens:", cached_tokens)
```
